```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset
X = np.array([2, 3, 4, 5, 6, 7, 8, 9, 10], dtype=float)
y = np.array([5.5, 6.0, 6.5, 7.0, 7.5, 8.0, 8.5, 9.0, 9.5], dtype=float)

# Reshape
X = X.reshape(-1, 1)

# Normalize X
X_mean = X.mean()
X_std = X.std()
X_scaled = (X - X_mean) / X_std

# Initialize parameters
w = 0
b = 0

learning_rate = 0.01
epochs = 1000

loss_history = []

# Gradient Descent
n = len(X_scaled)

for i in range(epochs):

    # Prediction
    y_pred = w * X_scaled.flatten() + b

    # Error
    error = y_pred - y

    # MSE
    loss = np.mean(error ** 2)
    loss_history.append(loss)

    # Gradients
    dw = (2/n) * np.sum(X_scaled.flatten() * error)
    db = (2/n) * np.sum(error)

    # Update
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Weight:", w)
print("Bias:", b)
print("Final MSE:", loss_history[-1])

# Plot loss
plt.plot(loss_history)
plt.xlabel("Iterations")
plt.ylabel("MSE Loss")
plt.title("Gradient Descent Loss")
plt.show()

# Prediction
study_hours = 7
study_scaled = (study_hours - X_mean) / X_std

prediction = w * study_scaled + b

print("Predicted CGPA:", prediction)
```
