```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# Data: [House Price, Annual Profit]
X = [[20, 5], [22, 6], [25, 7],
     [60, 18], [65, 20], [70, 22],
     [150, 50]]

# Apply DBSCAN
dbscan = DBSCAN(eps=12, min_samples=2)
labels = dbscan.fit_predict(X)

# Print cluster labels
print("Cluster Labels:", labels)

# Plot clusters
plt.scatter([x[0] for x in X],
            [x[1] for x in X],
            c=labels)

plt.title("DBSCAN: House Price vs Annual Profit")
plt.xlabel("House Price (₹ Lakhs)")
plt.ylabel("Annual Profit (₹ Lakhs)")
plt.show()
```
